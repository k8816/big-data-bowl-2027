```python
from __future__ import annotations

from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd

# Paths relative to repo root (notebook lives in notebooks/cursor_dev/)
REPO_ROOT = Path("../..").resolve()
IN_CSV = REPO_ROOT / "DerivedData" / "combine_players_dl.csv"

print("REPO_ROOT:", REPO_ROOT)
print("Input:", IN_CSV)
print("Exists:", IN_CSV.is_file())
```

## Load + drop missing forty

Drop rows with NaN `forty` before plotting.

```python
assert IN_CSV.is_file(), (
    f"Missing {IN_CSV}. Run combine_players_dl_export.ipynb first to build DerivedData."
)

df = pd.read_csv(IN_CSV)
print("Loaded shape:", df.shape)
print("Columns:", list(df.columns))

assert "forty" in df.columns, "CSV missing column: forty"

before = len(df)
plot_df = df.dropna(subset=["forty"]).copy()
print(f"Dropped {before - len(plot_df)} rows with NaN forty; remaining: {len(plot_df)}")
print(plot_df["forty"].describe())
```

## Histogram

Distribution of official electronic 40-yard dash times (seconds) for DL/edge prospects in the export.

```python
fig, ax = plt.subplots(figsize=(8, 5))
ax.hist(plot_df["forty"], bins=20, edgecolor="black", alpha=0.85)
ax.set_xlabel("Forty (seconds)")
ax.set_ylabel("Count")
ax.set_title("DL / edge forty-yard dash distribution")
ax.grid(axis="y", alpha=0.3)
fig.tight_layout()
plt.show()
```
